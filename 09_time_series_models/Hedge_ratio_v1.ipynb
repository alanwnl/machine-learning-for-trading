```python
# %%
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
from pykalman import KalmanFilter

# Define the stock pairs
# pairs = [
#     ('T', 'VOX'),
#     ('MDLZ', 'XOP'),
#     ('ERX', 'MDLZ'),
#     ('MDLZ', 'VDE'),
#     ('DIG', 'MDLZ'),
#     ('DJP', 'MDLZ'),
#     ('GSG', 'MDLZ'),
#     ('FXN', 'MDLZ'),
#     ('OIH', 'WBA'),
#     ('ERX', 'WBA')
# ]

pairs = [
    ('LMT', 'THD'),
    ('TIP', 'WBA'),
    ('CSX', 'HEDJ'),
    ('AGQ', 'DAL'),
    ('EPP', 'HPQ'),
    ('EWD', 'SWKS'),
    ('EPU', 'WBA'),
    ('EWA', 'HPQ'),
    ('BBY', 'TECL'),
    ('EWJ', 'STT'),
    ('SPY', 'XOM')
]

# Define the date range
start_date = '2016-12-01'
end_date = '2017-12-30'

# Helper function to get adjusted close series, handling MultiIndex and column variations
def get_adjusted_close(df, ticker):
    if df.empty:
        return pd.Series()  # Return empty series if no data
    if isinstance(df.columns, pd.MultiIndex):
        # Recent yfinance: MultiIndex, use 'Close' (adjusted)
        if 'Close' in df.columns.get_level_values(0):
            return df['Close'][ticker]
        else:
            raise KeyError(f"No 'Close' column found for {ticker}")
    else:
        # Older yfinance: flat columns
        if 'Adj Close' in df.columns:
            return df['Adj Close']
        elif 'Close' in df.columns:
            return df['Close']
        else:
            raise KeyError(f"No adjusted close column found for {ticker}")

# KFHedgeRatio function returning hedge ratio and intercept (fixed sign for positive correlation)
def KFHedgeRatio(x, y):
    """Estimate Hedge Ratio and Intercept using Kalman Filter"""
    delta = 1e-3
    trans_cov = delta / (1 - delta) * np.eye(2)
    obs_mat = np.expand_dims(np.vstack([[x], [np.ones(len(x))]]).T, axis=1)

    kf = KalmanFilter(n_dim_obs=1, n_dim_state=2,
                      initial_state_mean=[0, 0],
                      initial_state_covariance=np.ones((2, 2)),
                      transition_matrices=np.eye(2),
                      observation_matrices=obs_mat,
                      observation_covariance=2,
                      transition_covariance=trans_cov)

    state_means, _ = kf.filter(y.values)
    # Return hedge ratio and intercept (removed negative for alignment)
    # return state_means[:, 0], state_means[:, 1]
    return state_means[:, 0], state_means[:, 1]


def estimate_half_life(spread):
    X = spread.shift().iloc[1:].to_frame().assign(const=1)
    y = spread.diff().iloc[1:]
    beta = (np.linalg.inv(X.T @ X) @ X.T @ y).iloc[0]
    halflife = int(round(-np.log(2) / beta, 0))
    return max(halflife, 1)

# Function to download data, calculate KFHedgeRatio, and plot with estimated curve
def process_stock_pair(s1, s2, start, end):
    # Download data for both stocks with auto_adjust=True for adjusted prices
    stock1 = yf.download(s1, start=start, end=end, progress=False, auto_adjust=True)
    stock2 = yf.download(s2, start=start, end=end, progress=False, auto_adjust=True)
    
    # Check if data is available
    if stock1.empty or stock2.empty:
        print(f"No data available for pair ({s1}, {s2})")
        return None
    
    # Extract adjusted close prices
    adj_close1 = get_adjusted_close(stock1, s1)
    adj_close2 = get_adjusted_close(stock2, s2)
    
    # Create DataFrame
    data = pd.DataFrame({
        f'{s1}_Adj_Close': adj_close1,
        f'{s2}_Adj_Close': adj_close2
    }).dropna()
    
    # Calculate KFHedgeRatio and intercept
    if not data.empty:
        hedge_ratios, intercepts = KFHedgeRatio(data[f'{s1}_Adj_Close'], data[f'{s2}_Adj_Close'])
        data['KFHedgeRatio'] = hedge_ratios
        data['KFHedgeIntercept'] = intercepts
        # Calculate estimated curve: y = β * x + α
        
        data['Spread_FUCK'] = data[f'{s1}_Adj_Close'].add(data[f'{s2}_Adj_Close'].mul(hedge_ratios))
        data[f'{s2}_Estimated'] = hedge_ratios * data[f'{s1}_Adj_Close'] + intercepts
        # Calculate spread (residual)
        data['Spread'] = data[f'{s2}_Adj_Close'] -  data[f'{s2}_Estimated']
        data['Half_life'] = estimate_half_life(data['Spread']) 
    else:
        print(f"No valid data for KFHedgeRatio calculation for pair ({s1}, {s2})")
        return None
    return data

# Loop through each pair and process
for s1, s2 in pairs:
    data = process_stock_pair(s1, s2, start_date, end_date)
    print(data)
    print(data['Half_life'].describe())
     # Save DataFrame to CSV
    # filename = f'{s1}_{s2}_2016-12_to_2019-06.csv'
    # data.to_csv(filename)
    # print(f'Saved data with KFHedgeRatio, intercept, estimated curve, and spread for pair ({s1}, {s2}) to {filename}')
    
    # Create figure with three subplots (upper: prices + estimated curve, middle: hedge ratio, lower: spread)
    fig, (ax1, ax3, ax4) = plt.subplots(3, 1, figsize=(12, 12), height_ratios=[1, 1, 1])
    
    # Upper subplot: Adjusted closing prices and estimated curve
    # Plot first stock on primary y-axis
    ax1.plot(data.index, data[f'{s1}_Adj_Close'], color='blue', label=f'{s1} Adj Close')
    ax1.set_ylabel(f'{s1} Adjusted Close Price ($)', color='blue')
    ax1.tick_params(axis='y', labelcolor='blue')
    ax1.grid(True, alpha=0.3)
    
    # Create secondary y-axis for second stock and estimated curve
    ax2 = ax1.twinx()
    ax2.plot(data.index, data[f'{s2}_Adj_Close'], color='red', label=f'{s2} Adj Close')
    # ax2.plot(data.index, data['Spread_FUCK'], color='purple', linestyle='--', label=f'{s2} Estimated (β * {s1} + α)')
    ax2.plot(data.index, data[f'{s2}_Estimated'], color='purple', linestyle='--', label=f'{s2} Estimated (β * {s1} + α)')
    ax2.set_ylabel(f'{s2} Adjusted Close Price ($)', color='red')
    ax2.tick_params(axis='y', labelcolor='red')
    
    # Upper subplot labels and title
    ax1.set_title(f'Adjusted Closing Prices and Estimated Curve for {s1} and {s2} (Dec 2016 - Jun 2019)')
    
    # Legend for upper subplot
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
    
    # Middle subplot: KFHedgeRatio
    ax3.set_title('Kalman Filter Hedge Ratio')
    ax3.plot(data.index, data['KFHedgeRatio'], color='green', label='KFHedgeRatio')
    ax3.set_xlabel('Date')
    ax3.set_ylabel('KFHedgeRatio', color='green')
    ax3.tick_params(axis='y', labelcolor='green')
    ax3.grid(True, alpha=0.3)
    ax3.legend(loc='upper left')
    
    
    # Lower subplot: Spread
    ax4.set_title('Kalman Filter Spread (Residual)')
    ax4.plot(data.index, data['Spread'], color='black', label='Spread')
    ax4.set_xlabel('Date')
    ax4.set_ylabel('Spread', color='black')
    ax4.tick_params(axis='y', labelcolor='black')
    ax4.grid(True, alpha=0.3)
    ax4.legend(loc='upper left')

    
    # Adjust layout to prevent overlap
    plt.tight_layout()
    
    # Display the plot
    plt.show()
    
```
