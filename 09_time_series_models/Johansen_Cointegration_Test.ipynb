```python
# Import libraries
import numpy as np
import pandas as pd
from pandas_datareader import data as pdr
import yfinance as yf
from statsmodels.tsa.vector_ar.vecm import coint_johansen
```

```python
# Set the stock list
stock_list = ['AAPL', 'AMZN', 'MSFT']

# Set the start date and the end date
start_date = '2010-01-01'
end_date = '2022-10-16'
```

```python
# Download the data with auto_adjust=False to include 'Adj Close'
data = yf.download(stock_list, start=start_date, end=end_date, auto_adjust=False)['Adj Close']

# Define stock pairs
stock_pairs = [('AAPL', 'AMZN'), ('MSFT', 'AAPL'), ('AMZN', 'MSFT')]

# Separate the output sections
print("\n" + "-" * 50 + "\n")

# Perform the Johansen Cointegration Test for each pair separately
for i, (stock1, stock2) in enumerate(stock_pairs):
    # Select data for the pair and drop missing values
    pair_data = data[[stock1, stock2]].dropna()
    
    # Perform the Johansen Cointegration Test (det_order=0: constant term, k_ar_diff=1)
    # Testing the null hypothesis of zero cointegrating relationships
    coint_test_result = coint_johansen(pair_data, det_order=0, k_ar_diff=1)
    
    # Extract the trace statistics (lr1) and max eigenvalue statistics (lr2)
    trace_stats = coint_test_result.lr1
    max_eig_stats = coint_test_result.lr2
    
    # Critical values for trace statistic
    crit_values = coint_test_result.cvt
    
    print(f"Pair {i + 1} ({stock1} and {stock2}):")
    print(f"Trace Statistic (r=0): {trace_stats[0]:.4f}")
    print(f"Max Eigen Statistic (r=0): {max_eig_stats[0]:.4f}")
    print(f"Critical Values (r=0) [90%, 95%, 99%]: {crit_values[0]}")
    
    # Cointegration assessment (using 5% significance level for trace statistic)
    if trace_stats[0] > crit_values[0, 1]:
        print("Conclusion: Reject null hypothesis - Evidence of at least one cointegrating relationship.")
    else:
        print("Conclusion: Fail to reject null hypothesis - No evidence of cointegration.")
    
    print("\n" + "-" * 50 + "\n")

print("Cointegration Assessment: Testing for Zero Cointegrating Relationships (Null Hypothesis)\n")
```

```python
# Set the stock pairs
stock_pairs = [('AAPL', 'AMZN'), ('MSFT', 'AAPL'), ('AMZN', 'MSFT')]

# Set the start date and the end date
start_date = '2010-01-01'
end_date = '2022-10-16'

# Download stock price data for unique stocks with auto_adjust=False to ensure 'Adj Close' is available
unique_stocks = list(set(stock for pair in stock_pairs for stock in pair))
data = yf.download(unique_stocks, start=start_date, end=end_date, auto_adjust=False)['Adj Close']

# Perform the Johansen Cointegration Test for each pair
for i, (stock1, stock2) in enumerate(stock_pairs):
    # Select data for the pair and drop missing values
    pair_data = data[[stock1, stock2]].dropna()
    
    # Perform the test
    coint_test_result = coint_johansen(pair_data, det_order=0, k_ar_diff=1)
    
    # Extract trace statistics and critical values
    tracevalues = coint_test_result.lr1
    critical_values = coint_test_result.cvt
    
    # Check for cointegration at 5% significance level
    if tracevalues[0] > critical_values[0, 1]:
        print(f"Pair {i + 1} ({stock1} and {stock2}) is cointegrated.")
    else:
        print(f"Pair {i + 1} ({stock1} and {stock2}) is not cointegrated.")
```

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.api import VAR

# 1. Create sample time series data
# In a real scenario, you would load your own data.
np.random.seed(42)
data = np.random.randn(200, 2).cumsum(axis=0)
df = pd.DataFrame(data, columns=['y1', 'y2'])

# 2. Select the lag order using an information criterion (e.g., AIC)
model = VAR(df)
# The maxlags parameter is the maximum number of lags to check.
lag_selection = model.select_order(maxlags=10)
selected_lags = lag_selection.aic

print(f"Selected lag order (AIC): {selected_lags}")

df
```

```python
# Use the lag order selected by AIC (already assigned to selected_lags in previous cell)
# The lag order p for the VECM is k-1 where k is the lag order for the VAR model.
# So, we use selected_lags - 1.
# The `det_order` is for the deterministic term. 0 means no constant, 1 means a constant.
# The `k_ar_diff` is the number of lags in the VECM.

johansen_test = coint_johansen(df, det_order=0, k_ar_diff=int(selected_lags) - 1)

# Print the trace statistic and critical values
print("\n--- Johansen Cointegration Test Results ---")
print("Trace Statistic:", johansen_test.lr1)
print("Critical Values (90%, 95%, 99%):", johansen_test.cvt)

# Helper function to interpret the results
def interpret_johansen(test_results):
    """
    Interprets the Johansen test by comparing trace statistics to critical values.
    """
    num_series = len(test_results.lr1)
    print("\n--- Interpretation ---")
    for i in range(num_series):
        r = i
        trace_stat = test_results.lr1[i]
        crit_val_95 = test_results.cvt[i, 1] # 95% critical value

        print(f"Testing H0: r <= {r}")
        print(f"Trace Statistic = {trace_stat:.4f}")
        print(f"95% Critical Value = {crit_val_95:.4f}")

        if trace_stat > crit_val_95:
            print(f"Result: Reject H0. There are more than {r} cointegrating relationships.\n")
        else:
            print(f"Result: Fail to reject H0. There are at most {r} cointegrating relationships.\n")
            break # Stop when you fail to reject for the first time

interpret_johansen(johansen_test)
```
